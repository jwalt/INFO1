## 1. Variablen und Magics

`%timeit` misst die Laufzeit eines Ausdrucks.


```python
x = 21
print("x =", x)

%timeit sum(range(1000))
```

## 2. Diagramm

`%matplotlib inline` zeigt Plots direkt im Notebook.


```python
%matplotlib inline
import matplotlib.pyplot as plt

x = range(10)
y = [i**2 for i in x]

plt.plot(x, y, marker="o")
plt.title("y = x^2")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.show()
```

## 3. Interaktive Elemente

Falls `ipywidgets` installiert ist, erscheint ein Slider.


```python
try:
    import ipywidgets as widgets
    from IPython.display import display

    slider = widgets.IntSlider(value=5, min=0, max=20, description="Wert:")
    label = widgets.Label(value="Wert: 5")

    def on_change(change):
        label.value = f"Wert: {change['new']}"

    slider.observe(on_change, names="value")
    display(widgets.VBox([slider, label]))
except ImportError:
    print("ipywidgets ist nicht installiert. Optional installieren mit: pip install ipywidgets")
```
