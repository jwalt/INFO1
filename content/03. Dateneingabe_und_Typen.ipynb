# Daten eingeben – und was dabei mit den Typen passiert

Bisher haben wir Werte fest in den Code geschrieben und mit `print()` ausgegeben. Jetzt wird es interaktiv: Wir lernen, wie ein Programm **Daten vom Benutzer entgegennimmt** – und stoßen dabei automatisch auf das Thema **Datentypen**.

---

## Die klassische Methode: `input()`

In „normalen" Python-Programmen (Skripten und interaktiver Konsole) nimmt man Benutzereingaben mit der Funktion `input()` entgegen:

```python
name = input("Wie heißt du? ")
print(f"Hallo, {name}!")
```

`input()` tut Folgendes:
1. Es zeigt den Text in den Klammern als **Aufforderung (Prompt)** an.
2. Es wartet, bis der Benutzer etwas eintippt und **Enter** drückt.
3. Es gibt die eingegebene Zeile **als Text (String)** zurück.

Führe die Zelle unten aus und probiere es aus!


````python
# Aufgabe: name eingeben
name = input("Wie heißt du? ")
print(f"Hallo, {name}! Schön, dich kennenzulernen.")
````

Auch in Jupyter funktioniert `input()` – mit einem kleinen Unterschied:

- In einem **Terminal-Skript** erscheint die Eingabeaufforderung direkt im Terminal.
- In **Jupyter** öffnet `input()` ein **Eingabefeld unter der Zelle**, in das du tippen und mit Enter bestätigen kannst.

Du kannst also ganz normal `input()` verwenden. Beim Ausführen der Zelle unten erscheint unter der Zelle ein Texteingabefeld.

> **Wichtig für Jupyter:** Wenn du eine Zelle mit `input()` ausführst, musst du die Eingabe wirklich eintippen und Enter drücken – sonst „wartet" die Zelle. Das Ergebnis hängt dann von deiner Eingabe ab.


````python
stadt = input("In welcher Stadt wohnst du? ")
print(f"{stadt} ist eine tolle Stadt!")
````

## Die große Falle: `input()` liefert immer Text!

Hier lauert der **häufigste Anfänger-Fehler**. `input()` gibt **immer eine Zeichenkette (String)** zurück – auch wenn du eine Zahl eingibst!

Rechne zum Beispiel mal mit einer Zahleneingabe:

```python
zahl = input("Gib eine Zahl ein: ")
print(zahl * 2)
```


````python
# Falle: input() liefert einen String, keine Zahl
zahl = input("Gib eine Zahl ein: ")
print("Eingabe:", zahl)
print("Datentyp:", type(zahl))
print("zahl * 2 ergibt:", zahl * 2)
````

Schau dir die Ausgabe an! Wenn du zum Beispiel `7` eingegeben hast:

- `zahl * 2` ergibt **`77`** – nicht `14`!
- Das passiert, weil `"7"` ein **Text** ist und `"7" * 2` den Text einfach **wiederholt**.

> **Merkregel:** Benutzereingaben sind **immer Text**. Wenn du damit rechnen willst, musst du sie erst in eine Zahl **umwandeln**.

Hier kommen die **Datentypen** ins Spiel – schauen wir uns diese genauer an!


## Datentypen in Python: die grundlegenden (elementaren) Typen

Ein **Datentyp** beschreibt, was für eine Art Wert eine Variable enthält – und was man damit anstellen kann. Die wichtigsten elementaren Typen in Python sind:

| Typ      | Name            | Beispiel            | Beschreibung                       |
|----------|-----------------|---------------------|------------------------------------|
| `int`    | Ganzzahl         | `42`, `-3`, `0`     | Ganze Zahlen (ohne Nachkommastellen) |
| `float`  | Gleitkommazahl   | `3.14`, `2.0`       | Zahlen mit Nachkommastellen        |
| `str`    | Zeichenkette     | `"Hallo"`, `'a'`    | Text                               |
| `bool`   | Wahrheitswert    | `True`, `False`     | Ja/Nein-Wert                       |
| `NoneType` | Leerer Wert    | `None`              | „kein Wert" / nichts               |

Mit `type(wert)` kannst du herausfinden, welchen Typ eine Variable hat.

Schreibst du Daten direkt in den Programmcode, wählt Python automatisch den passenden Typen. Führe die Zelle unten aus:


````python
# Die elementaren Datentypen
a = 42              # int
b = 3.14            # float
c = "Hallo"         # str
d = True            # bool
e = None            # NoneType

print(type(a), "->", a)
print(type(b), "->", b)
print(type(c), "->", c)
print(type(d), "->", d)
print(type(e), "->", e)
````

### Expressions mit verschiedenen Typen

Je nach Typ verhält sich derselbe Operator *unterschiedlich*. Das ist wichtig:

```python
3 + 3      # int + int   -> 6 (Addition)
"3" + "3"  # str + str   -> "33" (Verkettung/aneinanderhängen)
3 * 3      # int * int   -> 9 (Multiplikation)
"3" * 3    # str * int   -> "333" (Text wiederholen)
```

Führe die Zelle unten aus – gleiche Symbole, aber je nach Datentyp komplett andere Ergebnisse!


````python
# Derselbe Operator, verschiedene Ergebnisse - je nach Datentyp
print('3 + 3   =', 3 + 3)
print('"3" + "3" =', "3" + "3")
print('3 * 3   =', 3 * 3)
print('"3" * 3 =', "3" * 3)

# Achtung: int + str funktioniert NICHT
# Diese Zeile wuerde einen Fehler werfen:
# print(3 + "3")
````

## Typumwandlung

Um aus einer Eingabe (String) eine Zahl zu machen, verwendest du die **Umwandlungsfunktionen**:

| Funktion     | Wandelt um in  | Beispiel                 |
|--------------|----------------|--------------------------|
| `int(wert)`  | Ganzzahl       | `int("7")` → `7`         |
| `float(wert)`| Gleitkommazahl | `float("7.5")` → `7.5`   |
| `str(wert)`  | Zeichenkette   | `str(42)` → `"42"`       |
| `bool(wert)` | Wahrheitswert  | `bool(1)` → `True`       |


````python
# Eingabe sofort in eine Zahl umwandeln
eingabe = input("Gib eine Zahl ein: ")
zahl = float(eingabe)   # String -> Gleitkommazahl
print("Du hast", eingabe, "eingegeben.")
print("Das Doppelte ist:", zahl * 2)
````

### Die häufigsten Fallstricke der Typumwandlung

1. **`int()` kann keine Kommazahlen-Strings** – `int("7.5")` wirft einen **Fehler**!
   ```python
   # int("7.5")  -> ValueError
   float("7.5")  # -> 7.5  (das funktioniert)
   ```

2. **`int()` rundet nicht ab bei float-Eingabe mit Komma im String.** Verwende `float()`.

3. **`int()` kann nur „saubere" Zeichenketten** – `int("7abc")` wirft einen **Fehler**.

4. **Text lässt sich nicht in eine Zahl umwandeln** – `int("Hallo")` → **Fehler**.

5. **Boolesche Werte sind eigentlich Zahlen** – `True` ist `1` und `False` ist `0`. Das führt oft zu Überraschungen bei Rechnungen.

Führe die Zelle unten aus, um ein paar dieser Fälle zu sehen:


````python
# Fallstricke der Typumwandlung - funktionierende Beispiele
print(int("7"))          # -> 7
print(float("7.5"))      # -> 7.5
print(float(3))          # int -> float: 3.0
print(int(3.99))         # float -> int (schneidet ab!): 3
print(str(42))           # -> "42"
print(True + True)       # bool verhaelt sich wie 1+1 -> 2

# Diese Zeilen wuerden einen Fehler werfen - kommentiere eine zum Testen aus:
# print(int("7.5"))     # ValueError
# print(int("Hallo"))   # ValueError
````

## Zahlendarstellungen: Hexadezimal, Oktal, Binär & Exponential

Bisher haben Zahlen „normal" (dezimal) ausgesehen, z. B. `42`. Python kann Zahlen aber in **verschiedenen Zahlensystemen** darstellen und mit **wissenschaftlicher (exponentialer) Notation** umgehen.

### Schreibweisen für verschiedene Zahlensysteme

Im Quellcode kannst du eine Zahl in einem anderen System als Zeichenkette *eingeben*:

| Präfix | System     | Beispiel | Dezimalwert |
|--------|------------|----------|-------------|
| `0x`   | Hexadezimal | `0xFF`  | `255`       |
| `0o`   | Oktal       | `0o10`  | `8`         |
| `0b`   | Binär       | `0b1010`| `10`        |

Python versteht diese Darstellungen direkt:


````python
# Zahlen in verschiedenen Systemen eingeben
print("0xFF (hex) =", 0xFF)
print("0b1010 (bin) =", 0b1010)
print("0o10 (oktal) =", 0o10)
````

### Hex/Oktal/Binär-Eingabe umwandeln: `int(string, base)`

Bisher haben wir Zahlen *im Quellcode* mit `0x`/`0o`/`0b` eingegeben. Was aber, wenn der Benutzer eine Hex-/Binär-Zahl **als Text eingibt** (z. B. über `input()`)?

Dann hilft die **zweite Form von `int()`**: `int(string, basis)` wandelt eine Zeichenkette, die in einem bestimmten Zahlensystem geschrieben ist, in eine **Dezimalzahl** um.

| Aufruf              | Basis | Ergebnis |
|---------------------|-------|----------|
| `int("FF", 16)`     | 16    | `255`    |
| `int("1010", 2)`    | 2     | `10`     |
| `int("17", 8)`      | 8     | `15`     |
| `int("42", 10)`     | 10    | `42`     |

> **Hinweis:** Mit der Basis `16` akzeptiert Python sowohl `"FF"` als auch `"0xFF"`. Wenn deine Eingabe das Präfix bereits enthält (z. B. `0xff` von `input()`), kannst du auch die **Basis `0`** verwenden – Python erkennt das System dann automatisch:
```python
int("0xff", 0)   # -> 255  (automatische Erkennung)
int("0b1010", 0) # -> 10
int("0o17", 0)   # -> 15
int("255", 0)    # -> 255
```


### Exponential- und Gleitkomma-Notation

Für sehr große oder sehr kleine Zahlen gibt es die **wissenschaftliche Notation** (Exponentialschreibweise). Der Buchstabe `e` steht für „mal 10 hoch":

```python
1e3    # = 1 * 10^3  = 1000.0
1.5e-2 # = 1.5 * 10^-2 = 0.015
```

Damit kann man z. B. ganz einfach die **Lichtgeschwindigkeit** (ca. 300 000 km/s) oder **sehr kleine Abstände** schreiben.


````python
# Exponential-Notation
print(1e3)          # 1000.0
print(1.5e-2)       # 0.015
print(6.022e23)     # Avogadro-Zahl (sehr gross)
print(9.11e-31)     # Elektronenmasse in kg (sehr klein)

# In Exponentialschreibweise ausgeben
zahl = 12345.678
print("normal:", zahl)
print("wissenschaftlich:", format(zahl, '.2e'))
````

### Zahlen formatieren (ausgeben)

Mit f-Strings kannst du festlegen, **wie** eine Zahl angezeigt wird:

```python
print(f"{zahl:08d}")   # 8 Stellen, mit Nullen aufgefuellt
print(f"{pi:.3f}")     # 3 Nachkommastellen
print(f"{zahl:+.2f}")  # mit Vorzeichen
```


````python
# Zahlen formatieren mit f-Strings
pi = 3.14159265
zahl = 42

print(f"pi mit 2 Nachkommastellen: {pi:.2f}")
print(f"pi als Exponential: {pi:.3e}")
print(f"Hex von 255: {255:x}")
print(f"{zahl:05d}")        # auf 5 Stellen auffuellen
print(f"{pi:10.2f}")        # Breite 10, 2 Nachkommastellen
````

## Exkurs: Explizite Typ-Annotationen (ein kleiner Teaser)

**Python ist eine „dynamisch typisierte" Sprache** – du musst den Typ einer Variable normalerweise **nicht** angeben. Python erkennt ihn automatisch:

```python
zahl = 7       # Python weiss: das ist ein int
```

Es gibt aber optional **Typ-Annotationen (Type Hints)**, mit denen du den erwarteten Typ **anschreiben** kannst:

```python
def begruesse(name: str) -> str:
    return f"Hallo, {name}!"

alter: int = 30
```

- `name: str` → die Variable/der Parameter *soll* ein `str` sein.
- `-> str` → die Funktion *gibt* einen `str` zurück.
- `alter: int = 30` → Variable `alter` mit erwartetem Typ `int`.

### Wann braucht man das?

- **Nicht zwingend** – Python läuft auch ganz ohne.
- **Sehr hilfreich bei größeren Projekten** – man sieht sofort, welcher Typ erwartet wird.
- **Zusammenarbeit** – andere Entwickler verstehen deinen Code besser.
- **Werkzeuge & Editoren** – moderne IDEs und Linter zeigen dir Fehler früh, bevor das Programm läuft.

### Was KANN das?

Die Annotationen sind **Hinweise**, keine harten Regeln. Python **erzwingt** den Typ zur Laufzeit **nicht** – du kannst trotzdem einen falschen Wert zuweisen. Aber Hilfswerkzeuge wie **mypy**, **Pyright** oder die Autovervollständigung deines Editors nutzen sie, um Fehler zu finden und dir besser zu helfen.


````python
# Typ-Annotationen in Aktion
def verdoppeln(zahl: int) -> int:
    return zahl * 2

ergebnis: int = verdoppeln(21)
print("Verdoppelte Zahl:", ergebnis)
print("Typ von 'ergebnis':", type(ergebnis))
````

## Zusammenfassung

- Mit `input("Frage: ")` nimmst du **Benutzereingaben** entgegen – **auch in Jupyter** (dort erscheint ein Eingabefeld unter der Zelle).
- **`input()` liefert immer einen String** – selbst bei Zahleneingabe.
- Mit `int()`, `float()`, `str()` und `bool()` kannst du **Typen umwandeln**.
- **Fallstricke:** `int("7.5")` und `int("Hallo")` werfen Fehler; Boolesche Werte sind intern `0`/`1`.
- Die elementaren Datentypen sind `int`, `float`, `str`, `bool` und `NoneType`.
- **Expressions** verhalten sich je nach Datentyp unterschiedlich (`3 + 3` → `6`, `"3" + "3"` → `"33"`).
- **Zahlendarstellungen:** Hexadezimal (`0x`), Oktal (`0o`), Binär (`0b`) und **Exponential-Notation** (`1e3`, `1.5e-2`).
- Mit `int(string, basis)` wandelst du **Hex-/Oktal-/Binär-Eingaben** in Dezimalzahlen um (z. B. `int("FF", 16)` → `255`).
- Mit f-Strings stellst du Zahlen in anderen Systemen dar oder formatierst sie (`{pi:.2f}`).
- **Type Hints** (`name: str`) sind optionale **Anmerkungen**, die den Code lesbarer machen und von Werkzeugen genutzt werden – Python erzwingt sie aber nicht.
