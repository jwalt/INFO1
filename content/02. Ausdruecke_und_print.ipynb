# Ausdrücke und Werte ausgeben

Im letzten Arbeitsblatt haben wir unser erstes Programm mit `print()` geschrieben. Jetzt schauen wir uns an, **wie Python Werte berechnen** kann – mit sogenannten **Ausdrücken (Expressions)** – und **wie wir diese Werte ausgeben**.

---

## Was ist ein Ausdruck (Expression)?

Ein **Ausdruck** ist ein Stück Code, das einen **Wert** ergibt. Zum Beispiel:

```python
1 + 1
```

Das ist ein Ausdruck – er ergibt den Wert `2`.

Python kann mit solchen Ausdrücken rechnen. Versuche es gleich selbst in der Zelle unten und drücke **Shift + Enter**.

> **Hinweis:** In Jupyter siehst du das Ergebnis des **letzten** Ausdrucks direkt unter der Zelle, auch ohne `print()`. Wir nutzen gleich `print()`, weil man damit mehr Möglichkeiten hat.


````python
# Ein einfacher Ausdruck
1 + 1
````

## Zahlen und Operatoren in Python

In Python kannst du mit Zahlen rechnen. Die wichtigsten **Rechenzeichen (Operatoren)** sind:

| Operator | Bedeutung        | Beispiel | Ergebnis |
|----------|------------------|----------|----------|
| `+`      | Addition         | `7 + 3`  | `10`     |
| `-`      | Subtraktion      | `7 - 3`  | `4`      |
| `*`      | Multiplikation   | `7 * 3`  | `21`     |
| `/`      | Division         | `7 / 3`  | `2.333…` |
| `//`     | Ganzzahlige Division | `7 // 3` | `2`   |
| `%`      | Rest (Modulo)    | `7 % 3`  | `1`      |
| `**`     | Potenz           | `7 ** 3` | `343`    |

> **Merke:** `/` ergibt immer eine **Gleitkommazahl** (mit Nachkommastellen), `//` schneidet den Rest ab und `%` gibt den Rest der Division zurück.

Probiere die folgenden Rechnungen aus. Denk dran: nur der **letzte** Ausdruck wird unter der Zelle angezeigt.

````python
# Ein paar Berechnungen
7 + 3
7 - 3
7 * 3
7 / 3
7 // 3
7 % 3
7 ** 3
````

## Mehrere Werte mit `print()` ausgeben

Mit `print()` kannst du mehrere Werte in **einem** Aufruf ausgeben, indem du sie durch **Kommas** trennst. Python fügt dann automatisch ein Leerzeichen zwischen den Werten ein:

```python
print("Das Ergebnis ist", 7 + 3)
```

#### Außerdem:
- Du kannst **Texte und Zahlen mischen** – Python wandelt die Zahlen automatisch in Text um.
- Du kannst **beliebige Ausdrücke** direkt in `print()` schreiben.

Führe die Zelle unten aus und betrachte die Ausgabe.


````python
# Mit print() mehrere Werte ausgeben
print("7 + 3 ergibt", 7 + 3)
print("Das Ergebnis von", 7, "*", 3, "ist", 7 * 3)
print("Hallo", "Welt", "mit", "mehreren", "Werten")
````

## Ausdrücke in Variablen speichern

Du kannst das Ergebnis eines Ausdrucks in einer **Variable** speichern und später wiederverwenden:

```python
ergebnis = 7 * 3
print("Das Ergebnis ist", ergebnis)
```

Das Gleichheitszeichen `=` **weist** der Variable auf der linken Seite den Wert auf der rechten Seite zu. Eine Variable ist also ein Speicherplatz, in dem du einen Wert aufbewahrst.


````python
# Ausdruck in einer Variable speichern
ergebnis = 7 * 3
print("Das Ergebnis ist", ergebnis)
````

## Übung: Rechne selbst

Ändere die Zahlen in der Zelle unten und führe sie erneut aus. Probiere auch `//`, `%` und `**` aus!


````python
# Deine eigenen Berechnungen
zahl_a = 12
zahl_b = 5

print("Addition:", zahl_a + zahl_b)
print("Multiplikation:", zahl_a * zahl_b)
print("Division:", zahl_a / zahl_b)
# ...
````

## Werte mit f-Strings ausgeben

Eine besonders flexible Art, Werte auszugeben, sind **f-Strings** – offiziell **„formatted string literals“**, auf Deutsch: **formatierte Zeichenkettenliterale**. Bei einem f-String stellst du vor das Anführungszeichen ein `f` und schreibst die Variablen oder Ausdrücke direkt in **geschweifte Klammern** `{}`:

```python
name = "Anna"
print(f"Hallo, {name}!")
```

#### Das ist der Vorteil:
- Du kannst Text, Variablen und Ausdrücke **direkt an der richtigen Stelle** mischen – ganz ohne Kommas zu zählen.
- Der Code ist dadurch oft **deutlich lesbarer** als viele `+` oder Komma-Verkettungen.

Beachte das `f` **direkt vor** dem Anführungszeichen. Alles in `{}` wird durch den tatsächlichen Wert ersetzt:


````python
# f-Strings zum Ausgeben
name = "Anna"
alter = 30

print(f"Hallo, {name}!")
print(f"{name} ist {alter} Jahre alt.")
````

Probiere dies wieder mit allen Rechenoperatoren aus:

````python
# Auch Ausdrücke direkt im f-String auswerten
zahl_a = 12
zahl_b = 5

print(f"{zahl_a} + {zahl_b} = {zahl_a + zahl_b}")
print(f"{zahl_a} geteilt durch {zahl_b} ist {zahl_a / zahl_b:.2f}")
````

## Warum `print()` verwenden, wenn Jupyter das Ergebnis eh zeigt?

Das ist eine gute Frage! In Jupyter siehst du das Ergebnis eines Ausdrucks direkt unter der Zelle. Aber:

- In **echten Skripten** (`.py`-Dateien) und Programmen wird nichts automatisch angezeigt – dort brauchst du `print()` unbedingt.
- Mit `print()` kannst du **mehrere Dinge in einer Zeile** ausgeben und **Beschriftungen** hinzufügen.
- Du möchtest meist genau kontrollieren, **was** ausgegeben wird und **wie**.

Deshalb ist `print()` eine der wichtigsten Funktionen überhaupt – wir werden sie in fast jedem Programm brauchen.

---

## Zusammenfassung

- Ein **Ausdruck (Expression)** ist ein Stück Code, das einen Wert ergibt (z. B. `1 + 1` → `2`).
- Die wichtigsten **Operatoren**: `+ - * / // % **`.
- In **Jupyter** erscheint das Ergebnis des **letzten** Ausdrucks einer Code-Zelle direkt unter der Zelle.
- Mit `print(wert1, wert2, ...)` gibst du mehrere Werte auf einmal aus.
- Ergebnisse kannst du in **Variablen** speichern: `ergebnis = 7 * 3`.
- Mit **f-Strings** (offiziell: *formatierte Zeichenkettenliterale*) kannst du Variablen direkt in Text einfügen: `f"Hallo, {name}!"`.
- In echten Skripten brauchst du `print()`, um Werte überhaupt zu sehen.
