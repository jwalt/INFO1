### Struktur von Notebooks

Dieses Notebook ist in **Jupyter** geschrieben, einer Umgebung, in der du Code direkt hier ausführen kannst. Ein Notebook besteht aus **Zellen**. Jede Zelle hat einen Typ, *Markdown* oder *Code*. Markdown ist für Texte wie dieser hier, während Code für Python-Code ist.

Markdown ist übrigens eine Text-Syntax, mit der einfache Formatierung möglich ist, z.B. **fett** oder *kursiv*, oder Überschriften wie die folgende.

## Dein erstes Programm: „Hello, World“

In Python kannst du Text auf den Bildschirm ausgeben, indem du den Befehl `print()` verwendest. Alles, was du in die Klammern schreibst, wird ausgegeben.

Gib in die Code-Zelle unten Folgendes ein und drücke **Shift + Enter**:

```python
print("Hello, world!")
```

````python
# Dies ist eine Code-Zelle. Man erkennt sie an der dunkleren Hintergrundfarbe. Schreibe hier dein Programm rein.
# Zeilen, die mit # beginnen, sind übrigens Kommentare und werden von Python ignoriert.
````

### Was ist hier passiert?

- `print` ist eine **Funktion**, die etwas auf dem Bildschirm ausgibt.
- In den Klammern `()` steht, was ausgegeben werden soll.
- Der Text `Hello, world!` ist eine **Zeichenkette (String)** und steht in doppelten Anführungszeichen `"` – oder auch in einfachen `'`.
