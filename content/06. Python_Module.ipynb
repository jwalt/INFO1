## 1) Was ist ein Modul?

Ein **Modul** ist im Grunde eine **normale Python-Datei** (Endung `.py`), die Funktionen, Variablen oder Klassen enthält. Man kann sie an anderer Stelle **importieren** und wiederverwenden.

Du hast `numpy` schon importiert:

```python
import numpy as np
```

`numpy` ist selbst ein Modul. Das gleiche Prinzip funktioniert mit **eigenen** Dateien: Du schreibst eine `.py`-Datei und importierst sie mit `import`.

**Vorteile der Modularisierung** (aus der Vorlesung):
- **Wiederverwendung** von Code an vielen Stellen.
- **Bessere Testbarkeit** – jede Funktion einzeln prüfbar.
- **Teamarbeit** wird erleichtert – mehrere Personen arbeiten an getrennten Modulen.
- Der Gesamtcode bleibt **lesbar** und strukturiert statt eines undurchschaubaren Gewirrs.

## 2) Eine eigene Python-Datei in JupyterLab anlegen

Jetzt legen wir **wirklich** eine eigene `.py`-Datei an. So geht es in **JupyterLab**:

1. Klicke oben in der Tab-Leiste auf das **Plus-Symbol** (`+`) oder wähle im Menü **Datei → Neu → Python File**.
2. Benenne die neue Datei z. B. in **`messtechnik.py`** um (Rechtsklick → *Umbenennen*).
3. Schreibe in diese Datei die Funktionen, die du später im Notebook verwenden möchtest.
4. Speichere die Datei (**Ctrl+S** / **Strg+S**).

> **Wichtig:** Die Datei muss im **gleichen Ordner** liegen wie dein Notebook, damit der `import` sie findet. In JupyterLab ist das meistens der aktive Arbeitsordner.

Als Beispiel legen wir die Datei `messtechnik.py` an. **Öffne/Schreibe diese Datei jetzt** (außerhalb dieses Notebooks) mit folgendem Inhalt:

```python

def kraft_zu_masse(kraft_N, g=9.81):
    """Rechnet eine Gewichtskraft (N) in eine Masse (kg) um."""
    return kraft_N / g

def mittelwert(messwerte):
    """Berechnet den Mittelwert einer Liste von Zahlen."""
    return sum(messwerte) / len(messwerte)
```


## 3) Das Modul importieren

Nun binden wir unser eigenes Modul `messtechnik` ein. Das funktioniert genauso wie bei `numpy`:

```python
import messtechnik
```

Danach kannst du auf die Funktionen mit `messtechnik.` zugreifen:

```python
masse = messtechnik.kraft_zu_masse(981)
```

> **Merkregel:** Importiere Module **am Anfang** des Notebooks (bzw. des Programms), bevor du sie verwendest.

````python
# Unser eigenes Modul importieren
import messtechnik

# Jetzt können wir die Funktionen nutzen
masse = messtechnik.kraft_zu_masse(981)
print("Masse zu 981 N Gewichtskraft:", masse, "kg")
````

## 5) Was passiert beim Import? `__name__`

Beim Import eines Moduls führt Python den **kompletten Code** der Datei aus. Manche Module haben aber **Test-Code** am Ende, der beim Import nicht unnötig laufen soll.

Dafür gibt es die `if __name__ == "__main__":`-Konstruktion:

```python
# messtechnik.py
if __name__ == "__main__":
    print("Dieses Modul wird direkt ausgefuehrt, nicht importiert.")
```

- Beim **Import** ist `__name__` gleich dem Modulnamen (z. B. `messtechnik`) → der Block läuft **nicht**.
- Wird die Datei **direkt ausgeführt** (z. B. als Skript), ist `__name__ == "__main__"` → der Block läuft.

Füge diesen Block am Ende deiner `messtechnik.py` hinzu. Wenn du das Modul importierst, wird der Test-Code **nicht** ausgeführt.


````python
# Test: Beim Import laeuft der __main__-Block NICHT
import messtechnik

print("Import abgeschlossen - kein Test-Output vom Modul oben.")
````

## 7) Zusammenfassung

- Ein **Modul** ist eine normale `.py`-Datei mit Funktionen/Variablen, die man **importieren** und **wiederverwenden** kann.
- In **JupyterLab** legst du sie über **Datei → Neu → Python File** an – im gleichen Ordner wie dein Notebook.
- Importieren: `import messtechnik`, dann `messtechnik.funktion()`. Optional mit Alias: `import messtechnik as mt`.
- Nach Änderungen an der `.py`-Datei muss das Modul **neu geladen** werden: entweder *Restart & Run All* oder `importlib.reload(messtechnik)`.
- Der Test-Code am Modulende wird durch `if __name__ == "__main__":` vor dem Import „geschützt".
- Die **Python-Standardbibliothek** (`math`, `random`, …) bietet viele fertige Module.
- Mit `from math import sqrt` importierst du einzelne Funktionen direkt.
