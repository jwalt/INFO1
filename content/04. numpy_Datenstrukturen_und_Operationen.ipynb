## 1) Was ist numpy? – Import

`numpy` ist ein **externes Paket**, das nicht automatisch verfügbar ist. Ein Paket ist eine Sammlung von Code-Bibliotheken (Modulen). Ein *externes* Paket gehört nicht zum normalen Lieferumfang von Python und muss separat nachinstalliert werden. In der Labor-Umgebung ist es aber schon installiert.

Außerdem muss man Module **importieren** um sie zu nutzen. Man kann dabei einen Alias-Namen vergeben. Bei `numpy` hat sich `np` eingebügert:

```python
import numpy as np
```

Ab jetzt können wir alle `numpy` Funktionen über `np.` aufrufen.

> **Wichtig in Jupyter:** `import numpy as np` muss **einmal** in einer Code-Zelle ausgeführt werden, bevor die folgenden Zellen funktionieren. Deshalb steht der Import in einer **eigenen Zelle** am Anfang.

````python
# numpy importieren
import numpy as np

# Hat der Import funktioniert? Diese Funktion gibt die Version aus:
print(np.__version__)
````

## 2) Warum überhaupt numpy?

Im Vergleich zu einer normalen Liste `[61.5, 64.2, 70.8]` hat ein numpy-Array mehrere Vorteile:

- **Effizientere Speichernutzung** – numpy-Arrays sind kompakt im Speicher, Listen sind das nicht.
- **Genauere Kontrolle über den Datentyp** – wir können festlegen, wie viele Bits ein Wert belegt (siehe Abschnitt 4).
- **Elementweise Berechnung** auf dem gesamten Array – ganz ohne `for`-Schleife.
- **Vektoren und Matrizen** mit vielen Dimensions-Operationen.

Führe die Zelle unten aus und vergleiche einen `type()`-Aufruf:

```python
liste = [61.5, 64.2, 70.8]
array = np.array([61.5, 64.2, 70.8])
```

````python
# Liste vs. numpy-Array
liste = [61.5, 64.2, 70.8]
array = np.array([61.5, 64.2, 70.8])

print("Typ der Liste: ", type(liste))
print("Typ des Arrays:", type(array))
````

## 4) numpy-Datentypen: explizite Größe

In der Vorlesung wurde erklärt: Jeder Wert hat einen Datentyp. Bei numpy kannst du den Typ **explizit mit einer Größe in Bit** angeben. Das gibt dir **Kontrolle über den Speicherbedarf** und den **Wertebereich**.

| numpy-Datentyp | Bedeutung                     | Beispiel     |
|----------------|-------------------------------|--------------|
| `np.int8`      | Ganzzahl, 8 Bit (mit Vorzeichen)  | `42`     |
| `np.int16`     | Ganzzahl, 16 Bit (mit Vorzeichen) | `42`     |
| `np.int32`     | Ganzzahl, 32 Bit (mit Vorzeichen) | `42`     |
| `np.int64`     | Ganzzahl, 64 Bit (mit Vorzeichen) | `42`     |
| `np.uint8`     | Ganzzahl, 8 Bit (ohne Vorzeichen) | `42`     |
| `np.float32`   | Gleitkommazahl (single precision) | `3.14`   |
| `np.float64`   | Gleitkommazahl (double precision) | `3.14`   |
| `np.bool_`     | Wahrheitswert                 | `True`   |

Warum ist das wichtig? **Je weniger Bits, desto kleiner der mögliche Wertebereich.**

- `np.int16` reicht von **−32 768** bis **+32 767**.
- `np.uint8` reicht von **0** bis **255**.
- Bei **float** gilt ähnliches für die Genauigkeit.

Führe die Zelle unten aus und beobachte, was passiert, wenn man den Wertebereich **überschreitet**.

````python
# numpy-Datentypen anwenden
g = np.float64(9.81)     # Gleitkommazahl, 64-Bit
n = np.uint16(70000)      # Ganzzahl, 16-Bit --> was passiert hier? Korrigiere das, wenn nötig

print("g  :", g, "Typ:", type(g))
print("n  :", n, "Typ:", type(n))
````

| `M1 @ M2` | Matrixmultiplikation |

## 7) Zusammenfassung

- **numpy** ist das Paket für numerische Algebra und muss mit `import numpy as np` geladen werden.
- **numpy-Arrays** (`np.ndarray`) sind **homogen** und effizient; sie können **1D, 2D, 3D, …** sein.
- Erzeugen: `np.array(...)`, `np.asarray(...)`, `np.ones(n)`, `np.zeros(n)`, `np.identity(n)`.
- **Dimensionen**: `a.shape` gibt `(Zeilen, Spalten)` zurück.
- **Zugriff**: `a[zeile][spalte]`, Index fängt bei **0** an.
- **Explizite Typen** (z. B. `np.int16`, `np.uint8`, `np.float64`) kontrollieren Speicher und **Wertebereich** – Überlauf führt zu Fehlern (Fallbeispiel Ariane 5).
- **Operationen**: `a * 2`, `a.sum()`, `a.mean()`, `A @ B` für Matrixmultiplikation.
