### 6) Zählergesteuerte Schleife: `for` mit `range()`

````python
# for + range: von 0 bis 4
for i in range(5):
    print(i)

# Auch 2-3 Argumente möglich:
# range(2, 8)   -> 2 ... 7
# range(0, 10, 2) -> 0,2,4,6,8 (Schrittweite)
````

### 8) Kopfgesteuerte Schleife: `while`

Die **`while`-Schleife** prüft die Bedingung **vor** dem Block (kopfgesteuert):

```python
wert = 0
while wert < 5:
    print(wert)
    wert = wert + 1
```

> **Achtung:** Vergisst du die Zählzeile (`wert = wert + 1`), läuft die Schleife **ewig** – eine **Endlosschleife**! In Jupyter kannst du das mit `Kernel -> Unterbrechen` abbrechen.

````python
# while-Schleife
wert = 0
while wert < 5:
    print("Durchlauf:", wert)
    wert = wert + 1

# Was passiert, wenn du 'wert = wert + 1' entfernst?
````

### 9) Schachtelung von Schleifen

Schleifen können **ineinander verschachtelt** werden. Für **jeden Durchlauf der äußeren Schleife** läuft die **innere Schleife vollständig** durch (Vorlesung: *Rasterfahrt eines Koordinatenmessgeräts*).

Das klassische Beispiel: ein **Koordinatenraster** – für jede Zeile `z` jede Spalte `s` bearbeiten.

```python
for z in range(3):
    for s in range(4):
        print(f"Zeile {z}, Spalte {s}")
```

````python
# Verschachtelte Schleifen: Koordinatenraster
for z in range(3):
    for s in range(4):
        print(f"Zeile {z}, Spalte {s}")
    print("---")  # Trennlinie nach jeder Zeile
````

### 13) Rekursive Funktionen

Eine Funktion kann **sich selbst aufrufen** – das nennt man **Rekursion**. Wichtig: Es braucht eine **Abbruchbedingung**, sonst entsteht eine **Endlosschleife**.

Das klassische Beispiel aus der Vorlesung ist die **Fakultät**:

```python
def fakul(n):
    """Berechnet n! rekursiv."""
    if n == 0:          # Abbruchbedingung (Basisfall)
        return 1
    return n * fakul(n - 1)   # rekursiver Aufruf
```

````python
# Rekursive Fakultät
def fakul(n):
    """Berechnet n! rekursiv."""
    if n == 0:
        return 1
    return n * fakul(n - 1)

print("fakul(0) =", fakul(0))   # 1
print("fakul(1) =", fakul(1))   # 1
print("fakul(4) =", fakul(4))   # 24
print("fakul(5) =", fakul(5))   # 120
````

## Teil D – Ausnahmen (Exceptions)

### 14) Was ist eine Ausnahme?

Eine **Ausnahme (Exception)** ist ein **Laufzeitfehler**, der das Programm beenden würde. Beispiel: Zugriff auf ein nicht existierendes Dictionary-Element:

```python
d = {"a": 1}
print(d["b"])    # KeyError!
```

````python
# Eine Ausnahme ausloesen
d = {"a": 1}
print(d["b"])    # KeyError - Programm wuerde hier abbrechen!
````

### 15) Ausnahmen mit `try`/`except` abfangen

Der **Try-Except-Mechanismus** fängt Ausnahmen ab und kehrt zur normalen Programmausführung zurück:

```python
try:
    gefaehrlicher_code()
except ValueError:
    print("Das hat nicht geklappt!")
```

Was passiert unten, wenn man `0` eingibt. Kannst du das auch abfangen?

````python
# try/except: Ausnahme abfangen
try:
    # int("7.5") wirft einen ValueError
    zahl = int(input("Gib eine ganze Zahl ein: "))
    print("Kehrwert von Zahl:", 1 / zahl)
except ValueError:
    print("Das war keine gueltige ganze Zahl!")
````

## Zusammenfassung

- **Abfragen:** `if`, `elif`, `else`; Einrückung ist **entscheidend**.
- **Logik:** `and`, `or`, `not` mit Kurzschluss-Verhalten.
- **Schleifen:** `for` mit `range()` (zählergesteuert), `for` über Listen, `while` (kopfgesteuert), verschachtelte Schleifen, `break`/`continue`.
- **Fußgesteuerte Schleife gibt es in Python nicht.**
