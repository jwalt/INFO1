````python
# Demo: Circle-Klasse definieren und nutzen
class Circle:
    def __init__(self, x, y, r):
        self.x = x
        self.y = y
        self.r = r

    def bewegen(self, dx, dy):
        self.x = self.x + dx
        self.y = self.y + dy

    def getArea(self):
        import math
        return math.pi * self.r ** 2

# Zwei Objekte (Instanzen) derselben Klasse
c1 = Circle(0, 0, 1.0)
c2 = Circle(5, 5, 2.0)

print("c1 Radius:", c1.r)
print("c2 Flaeche:", c2.getArea())

c1.bewegen(10, -3)
print("c1 Position nach bewegen:", c1.x, c1.y)
print("Position c2 unveraendert:", c2.x, c2.y)
````

## 2) Kapselung und Zugriffskontrolle

In der Vorlesung hast du das **Black-Box-Modell** kennengelernt: Der direkte Zugriff von außen auf die internen Daten wird unterbunden, alles läuft über eine **kontrollierte Schnittstelle**. So bleibt das Objekt **konsistent**.

In Python wird diese Sichtbarkeit über **Namenskonventionen** ausgedrückt (nicht erzwungen, sondern Disziplin!):

| Symbol (UML) | Python-Konvention | Bedeutung |
|---|---|---|
| `+` public | `name()` | von außen zugänglich |
| `#` protected | `_name()` (ein Unterstrich) | von außen nicht zugänglich |
| `-` private | `_Klasse__name()` (Klassenname + zwei Unterstriche, sehr unüblich) | nicht zugänglich |

In Python gelten **Attribute immer als protected** – man schreibt sie mit einem Unterstrich davor: `self._r`.

Zum **Lesen** und **Schreiben** stellt man **Getter** und **Setter** bereit. Im Setter wird die Wertprüfung zentral durchgeführt:

````python
# Demo: Setter/Getter mit Wertpruefung

class Circle:
    def __init__(self, x, y, r):
        self._x = x
        self._y = y
        self.setRadius(r)   # Setter auch im Konstruktor nutzen

    def setRadius(self, new_r):
        if new_r <= 0:
            raise ValueError("Radius muss > 0 sein!")
        self._r = new_r

    def getRadius(self):
        return self._r

# korrekt:
c = Circle(0, 0, 1.5)
print("Radius:", c.getRadius())

# Setter mit Pruefung -> Fehler bei ungueltigem Wert
# c.setRadius(-5)   # hier auskommentiert, sonst bricht die Zelle ab
````

## 3) Werte- und Referenzsemantik

In der Vorlesung ging es um den Unterschied:

- **Wertesemantik** (Kopie): Eine Zuweisung erzeugt eine **Kopie**. Änderungen an der Kopie wirken nicht auf das Original. → elementare Datentypen wie `int`, `float`.
- **Referenzsemantik** (Verweis): Zwei Namen zeigen auf **dasselbe** Objekt. Änderungen über einen Namen sind auch über den anderen sichtbar. → Objekte (und Listen).

Das ist wichtig beim Verständnis von Objekten und Beziehungen:

````python
# Demo: Wertesemantik vs. Referenzsemantik

# Wertesemantik: Kopie bei Zuweisung
a = 5
b = a
b = b + 1
print("a =", a, " b =", b)   # a bleibt 5

# Referenzsemantik: gleiches Objekt
class Punkt:
    def __init__(self, x):
        self.x = x

p1 = Punkt(5)
p2 = p1            # p2 ist KEIN neues Objekt, sondern verweist auf p1
p2.x = 99
print("p1.x =", p1.x)   # auch p1.x ist 99 geworden!
````

## 4) Objektbeziehungen

Objekte stehen selten allein. Die Vorlesung unterscheidet drei Beziehungen:

| Beziehung | Bedeutung | Lebensdauer der Teile |
|---|---|---|
| **Assoziation** | „benutzt" – loses Verhältnis | Teile unabhängig |
| **Aggregation** | „hat" – Teil-Ganzes | Teile können unabhängig existieren |
| **Komposition** | „besteht aus" – existenzabhängig | Teile existieren nur mit dem Ganzen |

In Python gibt es **keinen funktionalen Unterschied** zwischen Aggregation und Komposition (Objekte leben so lange, wie sie referenziert werden). Der Unterschied steckt im **Konzept**:

- **Aggregation:** Das Teil wird *außerhalb* erzeugt und an den Konstruktor übergeben.
- **Komposition:** Das Teil wird *innerhalb* des Konstruktors erzeugt.

````python
# Demo: Aggregation vs. Komposition

# Komposition: Raum wird IM Haus erzeugt
class Raum:
    def __init__(self, flaeche):
        self._flaeche = flaeche

class Haus:
    def __init__(self, gesamtflaeche, raeume):
        self._raeume = []
        raumflaeche = gesamtflaeche / raeume
        for r in range(raeume):
            # Komposition: Teile werden intern erzeugt
            self._raeume.append(Raum(raumflaeche))

haus = Haus(90, 3)

# Aggregation: Werkzeug existiert schon und wird uebergeben
class Werkzeug:
    def __init__(self, name):
        self._name = name

class Werkstatt:
    def __init__(self, werkzeug):
        self._werkzeug = werkzeug       # Aggregation: Teil von aussen uebergeben

mein_wz = Werkzeug("Schraubendreher")
ws = Werkstatt(mein_wz)
````

## 5) Vererbung

**Vererbung** ist eine *„ist ein"*-Beziehung: Eine **abgeleitete Klasse** (Sub-, Kindklasse) übernimmt Attribute und Methoden von der **Basisklasse** (Super-, Elternklasse). Sie kann

- geerbte Attribute/Methoden **erben**,
- neue **zufügen**,
- geerbte Methoden **überschreiben**.

In Python schreibst du die Basisklasse in Klammern: `class Quadrat(Viereck):`.

Ein besonderer Fall sind die **Konstruktoren**: In Python muss der Konstruktor der Basisklasse **explizit mit `super()`** aufgerufen werden.

````python
# Demo: Vererbung und super()
class Viereck:
    def __init__(self, breite, hoehe):
        self._breite = breite
        self._hoehe = hoehe

    def getArea(self):
        return self._breite * self._hoehe

    def __str__(self):
        return f"Rechteck {self._breite}x{self._hoehe}"


class Quadrat(Viereck):
    def __init__(self, seite):
        # Oberklassen-Konstruktor explizit aufrufen
        super().__init__(seite, seite)

    def __str__(self):
        return f"Quadrat {self._breite}x{self._breite}"


def zeige(v):
    print(v, "-> Flaeche:", v.getArea())

zeige(Viereck(3, 4))
zeige(Quadrat(5))
````

## 6) Polymorphie

**Polymorphie** (Vielgestaltigkeit) heißt: Objekte **unterschiedlicher Klassen** lassen sich über die **gemeinsame Basisklasse** verwalten, und es wird automatisch die **richtige Methode** des konkreten Objekts aufgerufen. An welcher Methode das ist, entscheidet sich erst **beim Aufruf** (*späte Bindung*) anhand des tatsächlichen Objekts.

In Python reicht es, wenn alle abgeleiteten Klassen dieselbe Methodensignatur (z. B. `getArea()`) besitzen. Bei einer **abstrakten Klasse** gibt die Basisklasse eine Methode nur *vor* (ohne Code), die Kindklassen müssen sie implementieren.

> **Duck-Typing:** Python löst Methoden ohnehin erst beim Aufruf auf – *„Wenn es wie eine Ente aussieht und wie eine Ente quakt, ist es eine Ente."* Polymorphie bleibt trotzdem empfehlenswert, weil sie Fehler früher erkennt. (Stichwort: Mehrfachvererbung/`diamond` vermeiden.)

Um in Python die Fehlererkennung durch explizite abstrakte Basisklassen nutzen zu können, gibt es das Paket `abc` mit der Klasse `ABC` (abstract base class) und einer sogenannten Annotation `@abstractmethod`. Diese schreibt man in eine Zeile direkt vor die Methode: 

````python
# Demo: Polymorphie mit abc - abstrakte Basisklasse
from abc import ABC, abstractmethod

class Figur(ABC):
    @abstractmethod
    def getArea(self):
        """Muss in jeder direkten oder indirekten Ableitung
        implementiert werden - sonst keine Instanz moeglich."""
        ...

class Rechteck(Figur):
    def __init__(self, b, h):
        self._b, self._h = b, h
    def getArea(self):
        return self._b * self._h

class Kreis(Figur):
    def __init__(self, r):
        self._r = r
    def getArea(self):
        import math
        return math.pi * self._r ** 2

# Figur()   # -> TypeError: abstrakt, nicht instanziierbar

# Ein Array/Liste mit Objekten, die getArea() garantiert haben
figuren = [Rechteck(3, 4), Kreis(1.0), Rechteck(2, 5)]
for f in figuren:
    print(f.getArea())   # spaete Bindung: richtige Methode pro Objekt
````

## 6.1) `abc` – abstrakte Klassen und Schnittstellen

Das Demo oben nutzt jetzt das Modul **`abc`** (Abstract Base Classes) aus der Standardbibliothek:
die Basisklasse erbt von `ABC` und markiert `getArea()` mit `@abstractmethod`. Das **erzwingt** den Vertrag,
den die Folie *„In Python standardmäßig nicht erzwungen"* beschreibt:

- Eine **abstrakte Klasse** (mindestens eine abstrakte Methode) kann **nicht instanziiert** werden.
- Eine Ableitung darf erst dann eine Instanz erzeugen, wenn sie **alle** abstrakten Methoden implementiert hat.
- Ein **Interface** (Schnittstelle) ist in `abc` eine Klasse mit **ausschließlich** abstrakten Methoden –
  eine reine Vorlage ohne Code und ohne Hierarchie-Zwang. Ein Interface ist also ein Spezialfall einer abstrakten Klasse.

```python
from abc import ABC, abstractmethod

class Figur(ABC):
    @abstractmethod
    def getArea(self) -> float: ...
    # konkrete Methoden wuerden hier auch erlaubt sein (abstrakte Klasse)

class InterfaceNurAbstrakte(ABC):   # = Interface im Sinne der Folie
    @abstractmethod
    def notabschalten(self) -> bool: ...
```

Achtung: Der Name `Figur` wird in manchen Zellen weiter unten wiederverwendet – führe die Zellen von oben nach unten aus.

## 7) Zusammenfassung

- **Klasse** = Bauplan, **Objekt** = konkrete Instanz. Objekte bündeln **Attribute** (Daten) und **Methoden** (Verhalten).
- **Kapselung / Information Hiding:** Attribute sind protected (`_name`), Zugriff über **Setter/Getter** mit zentraler Wertprüfung.
- Spezielle Methoden: `__init__` (Konstruktor), `__str__` (Textausgabe), `__eq__` (Gleichheit).
- **Objektbeziehungen:** Assoziation (nutzt), Aggregation (hat, Teile unabhängig), Komposition (besteht aus, Teile existenzabhängig).
- **Werte- vs. Referenzsemantik:** elementare Typen kopieren, Objekte verweisen auf dasselbe Objekt.
- **Vererbung:** „ist ein"-Beziehung; erben, ergänzen, überschreiben; Basisklassen-Konstruktor mit `super()` aufrufen.
- **Polymorphie:** Objekte unterschiedlicher Klassen über die gemeinsame Basisklasse verwalten; die richtige Methode wird zur Laufzeit (späte Bindung) gewählt.
